```python
import numpy as np
from sklearn.linear_model import LinearRegression
import pandas as pd
from sklearn.preprocessing import MinMaxScaler as MinMaxScaler, LabelEncoder
from sklearn.cluster import DBSCAN
data_file = pd.read_csv('P4AI_BT1.csv')
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
print(data_file)
```

```python
num_coluum = ['sepal.length', 'sepal.width', 'petal.length', 'petal.width']
for col in num_coluum:
    if col in data_file.columns:
        data_file[col] = data_file[col].fillna(data_file[col].mean())
if 'variety' in data_file.columns:
    data_file['variety'] = data_file['variety'].fillna(data_file['variety'].mode()[0])
print(data_file)
```

```python
filter_data = data_file[(data_file['sepal.length']>5) & (data_file['sepal.width']>3)]
print(filter_data)
```

```python
scaler = MinMaxScaler()
df_scale = data_file.copy()
df_scale[num_coluum] = scaler.fit_transform(df_scale[num_coluum])
print(df_scale.head())
```

```python
label = LabelEncoder()
data_file['variety_encoded'] = label.fit_transform(data_file['variety'])
print(data_file[['variety_encoded', 'variety']])
```

```python
x=data_file[num_coluum]
y = data_file['variety_encoded']
reg_model = LinearRegression()
reg_model.fit(x, y)
print(reg_model.coef_)
print(reg_model.intercept_)
print(reg_model.score(x, y))
```

```python
sampled_df = data_file.sample(frac=0.5,replace=True,random_state=42)
print(len(sampled_df))
print(sampled_df.head())
```

```python
data_file['sepal.length.bins'] = pd.cut(data_file['sepal.length'], bins=3,labels=['Low','Medium','High'])
print(data_file[['sepal.length.bins', 'sepal.length']].head(100))
```

```python
dbscan = DBSCAN(eps=0.1, min_samples=3)
clusters = dbscan.fit_predict(df_scale[num_coluum])

data_file['clusters'] = clusters
print(data_file[data_file['clusters']==-1].head())
```
