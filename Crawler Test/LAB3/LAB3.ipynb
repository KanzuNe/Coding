```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('classic')
df = pd.read_csv('AB_NYC_2019 (1).csv')

counts = df['neighbourhood_group'].value_counts()
x = counts.index
y = counts.values

fig, ax = plt.subplots()

ax.bar(x, y, width=1, edgecolor="white", linewidth=0.7)

ax.set(ylim=(0, max(y)+1))
plt.savefig("cau_1.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python

top_neighbourhoods = df['neighbourhood'].value_counts().head(10)

fig, ax = plt.subplots()
ax.bar(top_neighbourhoods.index, top_neighbourhoods.values, edgecolor="white")

plt.xticks(rotation=45)
plt.savefig("cau_2.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, ax = plt.subplots()
sns.boxplot(data=df, x='neighbourhood_group', y='availability_365', ax=ax)
plt.savefig("cau_3.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python


plt.figure(figsize=(10, 8))
sns.scatterplot(
    data=df,
    x='longitude',
    y='latitude',
    hue='neighbourhood_group',
    s=8,
    alpha=0.4
)
plt.savefig("cau_4.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python

corr = df.corr(numeric_only=True)
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.savefig("cau_5.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
import plotly.express as px

fig = px.scatter_geo(
    df.head(1000),
    lat="latitude",
    lon="longitude",
    color="price",
    hover_name="name",
    hover_data=["neighbourhood", "price"]
)
fig.show()
```
